## 1. Primer contacto con los conjuntos

* **a) ¿Por qué los valores repetidos aparecen una sola vez?:**  
  Porque un conjunto (`set`) en Python es una colección de elementos únicos. Cuando procesa los valores repetidos, calcula la función hash de cada uno y descarta automáticamente los duplicados para que ningún elemento se repita dentro de la estructura.

* **b) ¿Por qué un conjunto vacío se crea con `set()` y no con `{}`?:**  
  Porque en la sintaxis de Python las llaves vacías `{}` están reservadas exclusivamente para crear un diccionario vacío (`dict`). La única manera de indicarle al lenguaje que queremos inicializar un conjunto vacío es mediante el constructor `set()`.

* **c) ¿Sería correcto intentar mostrar `lenguajes[0]`? Justifiquen:**  
  No, no es correcto y produce un error (`TypeError: 'set' object is not subscriptable`). Los conjuntos son colecciones no ordenadas y sin índices: sus elementos no ocupan una posición secuencial en memoria, por lo que no admiten indexación ni rebanado (*slicing*).

***
### Ejercicio 1 (Python):

```python
def main():
    lenguajesProgramacion = {"Python", "Java", "C", "Python", "C"}
    coleccionSinElementos = set()

    print("Contenido del conjunto:", lenguajesProgramacion)
    print("Tipo de dato del contenedor vacío:", type(coleccionSinElementos))


main()
```

## 2. Pertenencia y operaciones entre conjuntos

* **a) Unión (`|`):** Agrupa todos los elementos que están en al menos uno de los dos conjuntos, eliminando automáticamente los duplicados.
* **b) Intersección (`&`):** Obtiene únicamente los elementos que están presentes al mismo tiempo en ambos conjuntos.
* **c) Diferencia (`-`):** Deja los elementos que pertenecen al primer conjunto pero que no están presentes en el segundo.
* **d) Diferencia simétrica (`^`):** Incluye los elementos que están en uno u otro conjunto, pero descarta los que están en ambos.
* **e) Operadores de pertenencia (`in` y `not in`):** Evalúan si un valor puntual pertenece o no a un conjunto devolviendo un booleano (`True` o `False`).

### Justificación teórica — Inciso f:
* **¿Por qué la diferencia no es una operación conmutativa?:**  
  Porque la diferencia depende del conjunto inicial que tomemos como referencia. `grupo_a - grupo_b` se queda solo con los elementos propios y exclusivos de `grupo_a`, mientras que `grupo_b - grupo_a` se queda solo con los exclusivos de `grupo_b`. Como los conjuntos resultantes son distintos ($A - B \neq B - A$), la operación no cumple con la propiedad conmutativa.

***
### Ejercicio 2 (Python):

```python
def main():
    matriculasComisionAlfa = {101, 102, 103, 104, 105}
    matriculasComisionBeta = {104, 105, 106, 107}

    # a) Union
    padronTotal = matriculasComisionAlfa | matriculasComisionBeta
    print("a) Unión (Alfa | Beta):", padronTotal)

    # b) Interseccion
    padronCompartido = matriculasComisionAlfa & matriculasComisionBeta
    print("b) Intersección (Alfa & Beta):", padronCompartido)

    # c) Diferencias en ambos sentidos
    exclusivosAlfa = matriculasComisionAlfa - matriculasComisionBeta
    exclusivosBeta = matriculasComisionBeta - matriculasComisionAlfa
    print("c) Diferencia (Alfa - Beta):", exclusivosAlfa)
    print("   Diferencia (Beta - Alfa):", exclusivosBeta)

    # d) Diferencia simetrica
    padronNoCompartido = matriculasComisionAlfa ^ matriculasComisionBeta
    print("d) Diferencia simétrica (Alfa ^ Beta):", padronNoCompartido)

    # e) Evaluacion de pertenencia
    estaPresenteAlfa = 103 in matriculasComisionAlfa
    estaAusenteBeta = 108 not in matriculasComisionBeta
    print("e) ¿103 in matriculasComisionAlfa?:", estaPresenteAlfa)
    print("   ¿108 not in matriculasComisionBeta?:", estaAusenteBeta)


main()
```

## 3. Métodos y relaciones entre conjuntos

* **`add()`:** Agrega un nuevo elemento al conjunto. Si ya existía, no hace nada ni duplica el valor.
* **`remove()` vs `discard()`:** `remove()` elimina un elemento existente, pero si no lo encuentra genera un error `KeyError`. Por el contrario, `discard()` intenta borrarlo de forma segura sin romper la ejecución del programa ni lanzar excepciones en caso de que no exista.
* **`issubset()` y `issuperset()`:** Métodos booleanos de comprobación. `issubset()` verifica si todos los elementos de un conjunto están incluidos dentro de otro, mientras que `issuperset()` verifica la relación inversa (si contiene a todos los del otro).
* **`copy()` y `clear()`:** `copy()` genera un duplicado independiente en memoria para poder manipularlo o vaciarlo con `clear()` sin perder ni alterar los elementos del conjunto original.

***
### Ejercicio 3 (Python):

```python
def main():
    herramientasDesarrollo = {"Python", "PostgreSQL", "Git", "Docker"}
    requisitosObligatorios = {"Python", "Git"}

    print("Conjunto inicial:", herramientasDesarrollo)

    # a) Agregar una tecnologia mediante add()
    herramientasDesarrollo.add("Linux")
    print("\na) Tras add('Linux'):", herramientasDesarrollo)

    # b) Eliminar elemento existente mediante remove()
    herramientasDesarrollo.remove("Docker")
    print("b) Tras remove('Docker'):", herramientasDesarrollo)

    # c) Intentar eliminar inexistente con remove() capturando la excepcion KeyError
    print("\nc) Intento de eliminar elemento inexistente con remove():")
    try:
        herramientasDesarrollo.remove("Kubernetes")
    except KeyError:
        print("   Captura exitosa: remove() genero la excepcion esperada (KeyError).")

    # d) Repetir operacion con discard() y comparar comportamiento
    herramientasDesarrollo.discard("Kubernetes")
    print("d) Tras discard('Kubernetes'): la operacion no genero ningun error.")

    # e) Verificar con issubset() si las obligatorias estan incluidas
    esSubconjuntoValido = requisitosObligatorios.issubset(herramientasDesarrollo)
    print("\ne) ¿requisitosObligatorios es subconjunto de herramientasDesarrollo?:", esSubconjuntoValido)

    # f) Verificar con issuperset() la relacion inversa
    esSuperconjuntoValido = herramientasDesarrollo.issuperset(requisitosObligatorios)
    print("f) ¿herramientasDesarrollo es superconjunto de requisitosObligatorios?:", esSuperconjuntoValido)

    # g) Vaciar copia con clear() preservando el original
    duplicadoHerramientas = herramientasDesarrollo.copy()
    duplicadoHerramientas.clear()
    print("\ng) Duplicado vaciado tras clear():", duplicadoHerramientas)
    print("   Conjunto original intacto:", herramientasDesarrollo)


main()
```
