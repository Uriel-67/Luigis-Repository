### Ejercicio 1

* **a) ¿Qué significa que la información persista en el tiempo?**

  Significa que los datos se conservan de forma permanente en un dispositivo de almacenamiento no volátil, lo que evita que se pierdan al cerrar el programa y permite utilizarlos en ejecuciones posteriores.
* **b) ¿Qué diferencias encuentran entre una variable en memoria y un archivo?**

  Las variables se almacenan en la memoria principal (RAM), la cual es volátil; sus datos se borran irremediablemente cuando el programa termina o el equipo se apaga. Un archivo, por el contrario, reside en un medio no volátil y retiene la información de manera persistente a lo largo del tiempo.
* **c) ¿Qué ventajas y limitaciones presenta almacenar los datos como texto?**

  * **Ventajas:** Al almacenar los datos exclusivamente como caracteres, el archivo resultante es universal, fácilmente legible y editable por humanos utilizando cualquier procesador de texto básico.
  
  * **Limitaciones:** La organización de la información debe ser definida y gestionada estrictamente por el programa, ya que no toda línea de un archivo de texto representa necesariamente un registro estructurado. Además, requiere procesos de conversión y validación cada vez que se leen los caracteres para transformarlos a otros tipos de datos (como números o booleanos).

### Ejercicio 2



* **`open()`**: Abre el archivo indicando su nombre, el modo de apertura (ej. `"wt"` para escritura) y la codificación (`encoding="utf-8"`).
* **`write()`**: Escribe texto dentro del archivo. Exige agregar manualmente el salto de línea (`\n`) si se desea separar en registros.
* **`with`**: Gestor de contexto que asegura el cierre automático del archivo al terminar el bloque, protegiéndolo incluso si ocurre un error.

### Ejercicio 3

```python
def crear_alumnos(modo="wt"):

    with open("alumnos.txt", modo, encoding="utf-8") as archivo:
        print(f"\n--- Iniciando carga de alumnos (Modo '{modo}') ---")
        
        while True:
            # 1. Ingreso y validación del legajo
            try:
                legajo = int(input("Ingrese el legajo (-1 para finalizar): "))
            except ValueError:
                print("Error: El legajo debe ser un valor numérico entero.")
                continue
            if legajo == -1:
                print("Carga finalizada.")
                break
                
            # Validación de legajo positivo
            if legajo <= 0:
                print("Error: El legajo debe ser un número positivo mayor a cero.")
                continue
                
            # 2. Ingreso y validación del nombre
            nombre = input("Ingrese nombre y apellido: ").strip()
            if not nombre:
                print("Error: El nombre no puede estar vacío.")
                continue
                
            # 3. Ingreso y validación de la comisión
            try:
                comision = int(input("Ingrese la comisión (1 a 10): "))
                if comision < 1 or comision > 10:
                    print("Error: La comisión debe estar entre 1 y 10.")
                    continue
            except ValueError:
                print("Error: La comisión debe tener un formato numérico válido.")
                continue
                
            # Grabación del registro con los campos separados por punto y coma
            registro = f"{legajo};{nombre};{comision}\n"
            archivo.write(registro)
            print("-> Alumno guardado con éxito.")
```

### Resultados de las pruebas con distintos modos de apertura

* **Al ejecutar nuevamente usando el modo `"wt"`:** Los datos de los alumnos cargados en la primera ejecución se eliminan. Esto ocurre porque el modo de escritura (`wt`) borra por completo el contenido del archivo si este ya existe, creando un archivo nuevo y en blanco antes de empezar a escribir.
* **Al repetir con el modo `"at"`:** Los datos anteriores se conservan de forma segura. El modo de agregado (`at` - append) abre el archivo posicionado al final del mismo, permitiendo que los nuevos registros se añadan a continuación sin sobrescribir ni perder la información previamente cargada.

### Ejercicio 4

```python
def mostrar_alumnos():
    conteo_comisiones = {}
    
    print("Procesando registros")
    try:
        with open("alumnos.txt", "rt", encoding="utf-8") as archivo:
            for numero_linea, linea in enumerate(archivo, start=1):
                

                linea_limpia = linea.rstrip("\n")
                

                campos = linea_limpia.split(";")
                

                if len(campos) != 3:

                    print(f"Error de formato en línea {numero_linea}: {linea_limpia}")
                    continue
                

                legajo_str, nombre, comision_str = campos
                
                try:
                    legajo = int(legajo_str)
                    comision = int(comision_str)
                except ValueError:
                    print(f"Error de datos (no numéricos) en línea {numero_linea}: {linea_limpia}")
                    continue
                
  
                print(f"Válido -> Legajo: {legajo}, Nombre: {nombre}, Comisión: {comision}")
                

                if legajo > 10000:
                    print(f"  *** Destacado: {nombre} tiene un legajo mayor a 10000 ({legajo})")
                

                conteo_comisiones[comision] = conteo_comisiones.get(comision, 0) + 1


        print("\nCantidad de alumnos por comisión")
        if conteo_comisiones:
            for comision, cantidad in conteo_comisiones.items():
                print(f"Comisión {comision}: {cantidad} alumnos")
        else:
            print("No se encontraron registros válidos para contar.")

    except FileNotFoundError:
        print("El archivo 'alumnos.txt' no existe. Asegurate de ejecutar crear_alumnos() primero.")
```

### Ejercicio 5

```python
def crear_productos():
    with open("productos.csv", "wt", encoding="utf-8") as archivo:
        archivo.write("codigo;descripcion;cantidad;precio\n")
        
        while True:
            codigo = input("Ingrese código (FIN para salir): ").strip()
            if codigo.upper() == "FIN":
                break
                
            descripcion = input("Ingrese descripción: ").strip()
            
            try:
                cantidad = int(input("Ingrese cantidad: "))
                precio = float(input("Ingrese precio: "))
                
                if cantidad < 0 or precio <= 0:
                    print("Error: Cantidad >= 0 y Precio > 0.")
                    continue
            except ValueError:
                print("Error: Ingrese números válidos.")
                continue
            
            archivo.write(f"{codigo};{descripcion};{cantidad};{precio}\n")

def procesar_productos():
    total_acumulado = 0.0
    try:
        with open("productos.csv", "rt", encoding="utf-8") as archivo:
            archivo.readline()
            
            for linea in archivo:
                campos = linea.rstrip("\n").split(";")
                
                if len(campos) != 4:
                    continue
                
                codigo, descripcion, cantidad_str, precio_str = campos
                cantidad = int(cantidad_str)
                precio = float(precio_str)
                
                importe = cantidad * precio
                total_acumulado += importe
                
                if cantidad < 5:
                    print(f"Alerta: '{descripcion}' tiene menos de 5 unidades ({cantidad}).")
                    
        print(f"Importe Total: ${total_acumulado:.2f}")
                
    except FileNotFoundError:
        print("El archivo no existe.")
```

### Ejercicio 7

```python
def crear_archivo():
    with open("entregas.csv", "wt", encoding="utf-8") as archivo:
        archivo.write("grupo;proyecto;responsable;avance\n")

def existe_grupo(grupo):
    try:
        with open("entregas.csv", "rt", encoding="utf-8") as archivo:
            archivo.readline()
            for linea in archivo:
                campos = linea.rstrip("\n").split(";")
                if len(campos) == 4 and campos[0] == str(grupo):
                    return True
        return False
    except FileNotFoundError:
        return False

def cargar_entregas():
    with open("entregas.csv", "at", encoding="utf-8") as archivo:
        while True:
            grupo_str = input("Ingrese número de grupo (0 para salir): ").strip()
            if grupo_str == "0":
                break
            
            try:
                grupo = int(grupo_str)
                if grupo <= 0:
                    print("Error: El grupo debe ser positivo.")
                    continue
            except ValueError:
                print("Error: Número inválido.")
                continue
            
            if existe_grupo(grupo):
                print("Error: El grupo ya existe.")
                continue
                
            proyecto = input("Proyecto: ").strip()
            responsable = input("Responsable: ").strip()
            
            if not proyecto or not responsable:
                print("Error: Los textos no pueden estar vacíos.")
                continue
                
            try:
                avance = float(input("Avance (0-100): "))
                if avance < 0 or avance > 100:
                    print("Error: Avance entre 0 y 100.")
                    continue
            except ValueError:
                print("Error: Avance inválido.")
                continue
                
            archivo.write(f"{grupo};{proyecto};{responsable};{avance}\n")

def mostrar_entregas():
    try:
        with open("entregas.csv", "rt", encoding="utf-8") as archivo:
            archivo.readline()
            for numero_linea, linea in enumerate(archivo, start=2):
                campos = linea.rstrip("\n").split(";")
                
                if len(campos) != 4:
                    print(f"Línea {numero_linea} omitida: registro mal formado.")
                    continue
                
                print(f"Grupo: {campos[0]} | Proyecto: {campos[1]} | Responsable: {campos[2]} | Avance: {campos[3]}%")
    except FileNotFoundError:
        print("El archivo no existe.")

def buscar_entrega(numero_buscado):
    try:
        with open("entregas.csv", "rt", encoding="utf-8") as archivo:
            archivo.readline()
            for linea in archivo:
                campos = linea.rstrip("\n").split(";")
                if len(campos) != 4:
                    continue
                
                if int(campos[0]) == numero_buscado:
                    print(f"Encontrado - Proyecto: {campos[1]} | Responsable: {campos[2]} | Avance: {campos[3]}%")
                    return
            print("Grupo no encontrado.")
    except FileNotFoundError:
        print("El archivo no existe.")

def informar_menor_50():
    try:
        with open("entregas.csv", "rt", encoding="utf-8") as archivo:
            archivo.readline()
            for linea in archivo:
                campos = linea.rstrip("\n").split(";")
                if len(campos) != 4:
                    continue
                
                if float(campos[3]) < 50:
                    print(f"Alerta: Grupo {campos[0]} ({campos[1]}) tiene avance menor al 50%.")
    except FileNotFoundError:
        print("El archivo no existe.")

def calcular_promedio():
    suma = 0.0
    cantidad = 0
    try:
        with open("entregas.csv", "rt", encoding="utf-8") as archivo:
            archivo.readline()
            for linea in archivo:
                campos = linea.rstrip("\n").split(";")
                if len(campos) != 4:
                    continue
                
                try:
                    suma += float(campos[3])
                    cantidad += 1
                except ValueError:
                    continue
        
        if cantidad > 0:
            print(f"Promedio general de avance: {suma / cantidad:.2f}%")
        else:
            print("No hay registros válidos para calcular el promedio.")
    except FileNotFoundError:
        print("El archivo no existe.")
```
