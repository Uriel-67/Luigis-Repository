1. 

a) Los caracteres literales representan el texto tal cual está escrito, mientras que los metacaracteres tienen un significado especial para definir reglas de búsqueda.

b) Hacer match significa comprobar que un texto cumple con la regla o estructura que armaste en el patrón.

c) El módulo re se usa en python para buscar, extraer, validar o cambiar textos usando expresiones regulares.

d) Conviene poner la r de raw string adelante para que python no interprete mal las barras invertidas y las lea de forma literal.


2. 
a) ^ y $ marcan el inicio y el fin exacto de la cadena. ej: ^texto$.

b) . es cualquier caracter menos un salto de linea. ej: p.to.

c) [ ] arma una clase o rango. ej: [a-z].

d) [^ ] es para negar y excluir lo que sigue. ej: [^0-9].

e) | da alternativas. ej: (perro|gato).

f) ( ) agrupa elementos en un bloque. ej: (ja){3}.

g) ?, *, + son para repeticiones (0 o 1, 0 o más, 1 o más). ej: go+l.

h) las llaves son para cantidades exactas o rangos. ej: {2,4}.

i) \ sirve para escapar un metacarácter y usarlo como símbolo literal. ej: $.


3. 
El primer formato pide la palabra PROG y 3 numeros. Validos: PROG123, PROG000. Invalidos: PROG1, prog12.

El segundo es una palabra de solo letras (sin acentos) de 3 a 10 de largo. Validos: Hola, Python. Invalidos: yo, programacion.

El tercero son 4 numeros, un guion y 4 numeros. Validos: 1234-5678, 0000-0000. Invalidos: 123-456, numeros.

El cuarto formato son las patentes (3 letras y 3 numeros, o 2 letras, 3 numeros y 2 letras). Validos: ABC123, AB123CD. Invalidos: abc123, 123ABC.

Los patrones construidos:

a) r"^[A-Z]{2}[0-9]{4}$"[cite: 3]

b) r"^[0-9]{5}$"

c) r"^$[0-9]+$"[cite: 3]

d) r"^[A-Z][A-Za-z]*$"

e) r"^[0-9]{2}/[0-9]{2}/[0-9]{4}$"
