## 1. Concepto y organización de un archivo

* **a) ¿Qué significa que la información persista en el tiempo?:**  
  Significa que los datos quedan guardados físicamente en un soporte no volátil (disco, memoria flash, etc.) y no se pierden al finalizar la ejecución del programa ni al apagar la computadora, quedando disponibles para ejecuciones futuras.

* **b) ¿Qué diferencias encuentran entre una variable en memoria y un archivo?:**  
  * **Variable en memoria (RAM):** Es efímera y volátil; vive únicamente mientras el script está corriendo y se borra al cerrar el proceso.  
  * **Archivo:** Reside en almacenamiento permanente, permite que la información sobreviva a lo largo del tiempo y se procesa secuencialmente registro a registro para no saturar la memoria RAM.

* **c) ¿Qué ventajas y limitaciones presenta almacenar los datos como texto?:**  
  * **Ventajas:** Es portable y legible directamente por personas; puede abrirse, inspeccionarse o corregirse con cualquier editor simple (como el Bloc de notas).  
  * **Limitaciones:** Ocupa más espacio de almacenamiento que un archivo binario y requiere convertir los tipos de datos al leerlos (por ejemplo, transformar cadenas numéricas con `int()` o `float()`).

## 2. Apertura, procesamiento y cierre

### Análisis de los modos de apertura:
* **`"rt"` (Lectura):** Abre el archivo para leer texto. Requiere obligatoriamente que el archivo exista previamente; si no se encuentra en la ruta, se genera un error `OSError` (`FileNotFoundError`).
* **`"wt"` (Escritura):** Abre el archivo para escribir texto. Si no existe lo crea en blanco; si ya existía, vacía o sobreescribe por completo todo su contenido anterior.
* **`"at"` (Agregado):** Abre el archivo para añadir texto. Si no existe lo crea, y si ya contiene información ubica el cursor al final para escribir nuevos registros sin borrar lo previo.

### Función de los componentes:
* **`open()`:** Establece el canal de comunicación con el sistema operativo para acceder al archivo según el modo indicado y retorna el manejador del archivo.
* **`write()`:** Graba una cadena de caracteres en el archivo. No agrega saltos de línea automáticamente, por lo que es necesario concatenar de forma explícita el carácter `\n`.
* **Estructura `with`:** Administra el contexto de forma segura garantizando que el archivo se cierre automáticamente al finalizar el bloque o ante cualquier excepción que ocurra dentro de él.

***
### Ejercicio 2 (Python):

```python
def main():
    try:
        with open("notas.txt", "wt", encoding="utf-8") as manejadorNotas:
            manejadorNotas.write("Primer registro\n")
            manejadorNotas.write("Segundo registro\n")
            manejadorNotas.write("Tercer registro\n")
        print("Archivo 'notas.txt' generado exitosamente con with.")
    except OSError as errorArchivo:
        print("Error al acceder o escribir en el archivo:", errorArchivo)


main()
```

## 3. Escritura de registros de texto

Desarrollo de la función `crear_alumnos()` para generar `alumnos.txt`:
* Formato de almacenamiento por renglón: `legajo;nombre;comision\n`
* Corte de carga por centinela con `legajo == -1` (sin utilizar `break`).
* Validación de legajo positivo, nombre no vacío y comisión en rango de 1 a 10.
* Manejo de errores con `try/except` frente a entradas no numéricas (`ValueError`) y problemas de archivo (`OSError`).

***
### Ejercicio 3 (Python):

```python
def pedirEntero(mensaje):
    """Solicita un valor entero por teclado validando con try/except."""
    esValido = False
    numero = 0
    while not esValido:
        texto = input(mensaje).strip()
        try:
            numero = int(texto)
            esValido = True
        except ValueError:
            print("Entrada inválida: debe ingresar un número entero.")
    return numero


def crear_alumnos(modoApertura="wt"):
    """Carga registros de alumnos y los graba en alumnos.txt.
    Permite parametrizar el modo de apertura ('wt' o 'at')."""
    nombreArchivo = "alumnos.txt"
    try:
        with open(nombreArchivo, modoApertura, encoding="utf-8") as archAlumnos:
            continuaCarga = True
            while continuaCarga:
                nroLegajo = pedirEntero("\nIngrese legajo (-1 para finalizar): ")

                if nroLegajo == -1:
                    continuaCarga = False
                elif nroLegajo <= 0:
                    print("Error: el legajo debe ser un entero positivo.")
                else:
                    # Validacion de nombre no vacio
                    apenom = input("Ingrese nombre y apellido: ").strip()
                    while apenom == "":
                        print("Error: el nombre no puede quedar vacío.")
                        apenom = input("Ingrese nombre y apellido: ").strip()

                    # Validacion de comision (entero del 1 al 10)
                    comisionValida = False
                    nroComision = 0
                    while not comisionValida:
                        nroComision = pedirEntero("Ingrese comisión (1 a 10): ")
                        if 1 <= nroComision <= 10:
                            comisionValida = True
                        else:
                            print("Error: la comisión debe estar comprendida entre 1 y 10.")

                    # Grabacion del registro delimitado por ';' con salto de linea
                    registro = str(nroLegajo) + ";" + apenom + ";" + str(nroComision) + "\n"
                    archAlumnos.write(registro)
                    print(f"Registro cargado correctamente: {apenom}")

        print(f"\nFinalizó la grabación en '{nombreArchivo}' (modo '{modoApertura}').")
    except OSError as errorArchivo:
        print("Error al acceder o escribir en el archivo:", errorArchivo)


def main():
    # Prueba inicial en modo 'wt'
    print("=== PRUEBA 1: Carga inicial en modo 'wt' ===")
    crear_alumnos("wt")

    # Segunda prueba en modo 'wt' (para observar sobreescritura)
    print("\n=== PRUEBA 2: Reejecución en modo 'wt' (sobrescritura) ===")
    crear_alumnos("wt")

    # Tercera prueba en modo 'at' (para observar agregado al final)
    print("\n=== PRUEBA 3: Ejecución en modo 'at' (agregado) ===")
    crear_alumnos("at")


main()
```

### Comparación de resultados entre modos `"wt"` y `"at"`:
* **Al volver a ejecutar con `"wt"`:** El modo `"wt"` trunca el archivo existente. Se borran todos los registros que se habían guardado previamente y el archivo conserva únicamente los ingresados durante esa última ejecución.
* **Al ejecutar con `"at"`:** El modo `"at"` respeta y conserva todos los datos anteriores intactos, posiciona el puntero al final de `alumnos.txt` y anexa los nuevos registros a continuación.

## 4. Lectura y procesamiento secuencial

Desarrollo de la función `mostrar_alumnos()` para leer `alumnos.txt` secuencialmente:
* Procesamiento registro por registro con `for linea in arch` para no cargar el archivo completo en memoria.
* Limpieza de fin de línea usando `rstrip("\n")` para no afectar espacios válidos del texto.
* Separación de campos mediante `.split(";")` y verificación estricta de que existan exactamente 3 columnas por línea.
* Muestra de registros válidos y filtro de legajos mayores a 10000.
* Conteo de alumnos por comisión mediante un diccionario acumulador.
* Identificación y reporte de líneas con errores o formato inválido sin abortar la ejecución del resto del archivo.

***
### Ejercicio 4 (Python):

```python
def mostrar_alumnos():
    nombreArchivo = "alumnos.txt"
    conteoComisiones = {}
    lineasInvalidas = []

    try:
        with open(nombreArchivo, "rt", encoding="utf-8") as archAlumnos:
            print("=== a) LISTADO DE ALUMNOS VÁLIDOS ===")
            print("=== b) ALUMNOS CON LEGAJO MAYOR A 10000 ===")
            
            for linea in archAlumnos:
                lineaLimpia = linea.rstrip("\n")
                campos = lineaLimpia.split(";")

                # Verificamos que tenga exactamente 3 campos antes de convertir
                if len(campos) == 3:
                    cadenaLegajo = campos[0].strip()
                    nombreAlumno = campos[1].strip()
                    cadenaComision = campos[2].strip()

                    try:
                        nroLegajo = int(cadenaLegajo)
                        nroComision = int(cadenaComision)

                        if nroLegajo > 0 and nombreAlumno != "" and 1 <= nroComision <= 10:
                            # a) Muestra de registro valido
                            print(f"[Válido] Legajo: {nroLegajo} | Nombre: {nombreAlumno} | Comisión: {nroComision}")

                            # b) Informar legajo mayor a 10000
                            if nroLegajo > 10000:
                                print(f"    --> Destacado (>10000): {nombreAlumno} ({nroLegajo})")

                            # c) Conteo por comision con diccionario
                            conteoComisiones[nroComision] = conteoComisiones.get(nroComision, 0) + 1
                        else:
                            lineasInvalidas.append(lineaLimpia)
                    except ValueError:
                        lineasInvalidas.append(lineaLimpia)
                else:
                    lineasInvalidas.append(lineaLimpia)

        # c) Mostrar resultados del conteo
        print("\n=== c) CANTIDAD DE ALUMNOS POR COMISIÓN ===")
        if len(conteoComisiones) > 0:
            for nroComision in sorted(conteoComisiones.keys()):
                print(f"Comisión {nroComision}: {conteoComisiones[nroComision]} alumnos")
        else:
            print("No se registraron comisiones válidas.")

        # d) Reporte de lineas invalidas
        print("\n=== d) LÍNEAS CON FORMATO INVÁLIDO O ERRORES ===")
        if len(lineasInvalidas) > 0:
            for regFalla in lineasInvalidas:
                print(f"Línea omitida por error: '{regFalla}'")
        else:
            print("Todos los registros leídos presentaron un formato correcto.")

    except OSError as errorArchivo:
        print("Error al intentar leer el archivo:", errorArchivo)


def main():
    mostrar_alumnos()


main()
```

## 5. Archivos CSV y encabezados

Creación y lectura de `productos.csv` delimitado por punto y coma (`;`), **sin utilizar el módulo CSV** de Python:
* Estructura de cabecera: `codigo;descripcion;cantidad;precio`.
* Carga iterativa que finaliza al tipear el código `FIN` (el cual no se graba en el archivo).
* Omisión controlada del encabezado durante la lectura secuencial.
* Validaciones: cantidad $\ge 0$, precio $> 0$ y consistencia de datos numéricos.
* Cálculo del importe total y reporte de stock crítico (cantidad $< 5$).

***
### Ejercicio 5 (Python):

```python
def pedirEnteroPositivoOCero(mensaje):
    esValido = False
    numero = 0
    while not esValido:
        texto = input(mensaje).strip()
        try:
            numero = int(texto)
            if numero >= 0:
                esValido = True
            else:
                print("Error: El número debe ser mayor o igual a 0.")
        except ValueError:
            print("Error: Ingrese un número entero válido.")
    return numero


def pedirFlotantePositivo(mensaje):
    esValido = False
    numero = 0.0
    while not esValido:
        texto = input(mensaje).strip()
        try:
            numero = float(texto)
            if numero > 0:
                esValido = True
            else:
                print("Error: El precio debe ser estrictamente mayor a 0.")
        except ValueError:
            print("Error: Ingrese un valor numérico decimal válido.")
    return numero


def cargar_productos():
    nombreArchivo = "productos.csv"
    try:
        with open(nombreArchivo, "wt", encoding="utf-8") as archProductos:
            # Grabacion obligatoria del encabezado inicial
            archProductos.write("codigo;descripcion;cantidad;precio\n")
            
            continuaCarga = True
            while continuaCarga:
                codigo = input("\nIngrese código del producto ('FIN' para cortar): ").strip()
                
                if codigo.upper() == "FIN":
                    continuaCarga = False
                elif codigo == "":
                    print("Error: El código no puede estar vacío.")
                else:
                    descripcion = input("Ingrese descripción: ").strip()
                    while descripcion == "":
                        print("Error: La descripción no puede estar vacía.")
                        descripcion = input("Ingrese descripción: ").strip()

                    cantidad = pedirEnteroPositivoOCero("Ingrese stock disponible: ")
                    precio = pedirFlotantePositivo("Ingrese precio unitario: ")

                    registro = f"{codigo};{descripcion};{cantidad};{precio}\n"
                    archProductos.write(registro)
                    print(f"Producto '{descripcion}' registrado.")

        print(f"\nArchivo '{nombreArchivo}' creado con éxito.")
    except OSError as errorArchivo:
        print("Error al escribir el archivo de productos:", errorArchivo)


def procesar_productos():
    nombreArchivo = "productos.csv"
    totalGeneral = 0.0
    esPrimeraLinea = True

    try:
        with open(nombreArchivo, "rt", encoding="utf-8") as archProductos:
            print("\n=== DETALLE DE PRODUCTOS E IMPORTES ===")
            print("=== PRODUCTOS CON STOCK CRÍTICO (< 5) ===")

            for linea in archProductos:
                if esPrimeraLinea:
                    # Omitimos el encabezado de forma controlada
                    esPrimeraLinea = False
                else:
                    lineaLimpia = linea.rstrip("\n")
                    campos = lineaLimpia.split(";")

                    if len(campos) == 4:
                        cod = campos[0].strip()
                        desc = campos[1].strip()
                        cadCantidad = campos[2].strip()
                        cadPrecio = campos[3].strip()

                        try:
                            cant = int(cadCantidad)
                            prec = float(cadPrecio)

                            if cant >= 0 and prec > 0:
                                importeSubtotal = cant * prec
                                totalGeneral += importeSubtotal
                                print(f"Cod: {cod} | {desc} | Cant: {cant} | Precio: ${prec:.2f} | Subtotal: ${importeSubtotal:.2f}")

                                if cant < 5:
                                    print(f"    --> ALERTA DE STOCK CRÍTICO: {desc} tiene solo {cant} unidades.")
                        except ValueError:
                            print(f"Error en datos numéricos del registro: '{lineaLimpia}'")
                    else:
                        print(f"Registro con cantidad incorrecta de campos: '{lineaLimpia}'")

            print(f"\nIMPORTE TOTAL GENERAL DE STOCK: ${totalGeneral:.2f}")

    except OSError as errorArchivo:
        print("Error al abrir o leer el archivo de productos:", errorArchivo)


def main():
    cargar_productos()
    procesar_productos()


main()
```

## 6. Manejo de excepciones

### Comparación técnica: `try-except-finally` vs `with`

* **Uso de `with` (Recomendado):** Actúa como administrador de contexto. Se encarga de invocar de manera automática y garantizada al método `.close()` al salir del bloque identado, ya sea por finalización normal o ante el surgimiento imprevisto de cualquier excepción. Hace el código mucho más conciso, limpio y previene fugas de recursos del sistema operativo.
* **Uso de `try-except-finally`:** Exige declarar e inicializar la variable del archivo en `None` antes del bloque `try`. En la sección `finally`, es obligatorio verificar mediante una condición `if arch is not None:` que el archivo realmente se haya logrado abrir antes de llamar a `.close()`, ya que si la función `open()` falló arrojando un error, la variable no contendrá ningún manejador y llamar a `.close()` causaría un nuevo error no controlado.

***
### Ejercicio 6 (Python):

```python
def procesarArchivoSeguro(nombreArchivo):
    """Demostración de lectura controlada capturando OSError y ValueError explícitamente."""
    try:
        with open(nombreArchivo, "rt", encoding="utf-8") as arch:
            esEncabezado = True
            for linea in arch:
                if esEncabezado:
                    esEncabezado = False
                else:
                    lineaLimpia = linea.rstrip("\n")
                    partes = lineaLimpia.split(";")
                    
                    if len(partes) != 4:
                        print(f"Aviso: Registro con cantidad inesperada de campos omitido -> '{lineaLimpia}'")
                    else:
                        try:
                            cantidad = int(partes[2].strip())
                            precio = float(partes[3].strip())
                            print(f"Procesado OK: {partes[1].strip()} -> Unidades: {cantidad}, Precio: ${precio:.2f}")
                        except ValueError:
                            print(f"Aviso: El registro contiene valores numéricos incompatibles -> '{lineaLimpia}'")
    except OSError as errorAcceso:
        print("No se pudo acceder al archivo:", errorAcceso)


def main():
    print("=== Procesando productos.csv con control de excepciones ===")
    procesarArchivoSeguro("productos.csv")

    print("\n=== Prueba con un archivo inexistente ===")
    procesarArchivoSeguro("archivo_fantasma.csv")


main()
```

## 7. Desafío aplicado al Proyecto Integrador

Gestión integral de entregas mediante el archivo `entregas.csv`:
* Campos por registro: `numero_grupo;nombre_proyecto;integrante_responsable;avance`
* Modularización de tareas en funciones independientes:
  * `crear_entregas_csv()`: Escribe la fila de encabezado único en modo `"wt"`.
  * `grupo_ya_existe(nroGrupo)`: Verifica secuencialmente que no se dupliquen grupos.
  * `cargar_entrega()`: Valida número de grupo positivo, cadenas no vacías y avance entre 0 y 100.
  * `mostrar_entregas()`: Muestra todos los registros bien conformados y omite los mal formados.
  * `buscar_entrega(nroGrupo)`: Búsqueda secuencial por número de grupo.
  * `proyectos_avance_menor_50()`: Reporta proyectos con porcentaje de avance $< 50\%$.
  * `calcular_promedio_avance()`: Calcula la media sin riesgo de división por cero.

***
### Ejercicio 7 (Python):

```python
RUTA_ENTREGAS = "entregas.csv"


def crear_entregas_csv():
    """Crea el archivo inicializando únicamente el encabezado."""
    try:
        with open(RUTA_ENTREGAS, "wt", encoding="utf-8") as arch:
            arch.write("numero_grupo;nombre_proyecto;integrante_responsable;porcentaje_avance\n")
        print(f"Archivo '{RUTA_ENTREGAS}' inicializado correctamente con su encabezado.")
    except OSError as error:
        print("Error al inicializar el archivo de entregas:", error)


def grupo_ya_existe(nroGrupoBuscado):
    """Recorre secuencialmente el archivo para chequear si el grupo ya fue cargado."""
    existe = False
    try:
        with open(RUTA_ENTREGAS, "rt", encoding="utf-8") as arch:
            esEncabezado = True
            for linea in arch:
                if esEncabezado:
                    esEncabezado = False
                else:
                    campos = linea.rstrip("\n").split(";")
                    if len(campos) == 4:
                        try:
                            grupoActual = int(campos[0].strip())
                            if grupoActual == nroGrupoBuscado:
                                existe = True
                        except ValueError:
                            pass
    except OSError:
        existe = False
    return existe


def cargar_entrega():
    """Solicita y valida los datos de una entrega para agregarlos al archivo."""
    print("\n--- CARGA DE NUEVA ENTREGA ---")
    
    # 1. Validar numero de grupo positivo y no repetido
    nroGrupoValido = False
    nroGrupo = 0
    while not nroGrupoValido:
        textoGrupo = input("Ingrese número de grupo: ").strip()
        try:
            nroGrupo = int(textoGrupo)
            if nroGrupo <= 0:
                print("Error: El número de grupo debe ser mayor a 0.")
            elif grupo_ya_existe(nroGrupo):
                print(f"Error: El grupo {nroGrupo} ya se encuentra registrado en el archivo.")
            else:
                nroGrupoValido = True
        except ValueError:
            print("Error: Ingrese un número entero.")

    # 2. Validar cadenas no vacias
    nomProyecto = input("Nombre del proyecto: ").strip()
    while nomProyecto == "":
        print("Error: El nombre del proyecto no puede estar vacío.")
        nomProyecto = input("Nombre del proyecto: ").strip()

    responsable = input("Integrante responsable: ").strip()
    while responsable == "":
        print("Error: El responsable no puede quedar vacío.")
        responsable = input("Integrante responsable: ").strip()

    # 3. Validar porcentaje de avance (0 a 100)
    avanceValido = False
    avance = 0.0
    while not avanceValido:
        textoAvance = input("Porcentaje de avance (0 a 100): ").strip()
        try:
            avance = float(textoAvance)
            if 0.0 <= avance <= 100.0:
                avanceValido = True
            else:
                print("Error: El porcentaje debe encontrarse entre 0 y 100.")
        except ValueError:
            print("Error: Ingrese un valor numérico válido.")

    # 4. Grabacion en modo append ('at')
    try:
        with open(RUTA_ENTREGAS, "at", encoding="utf-8") as arch:
            registro = f"{nroGrupo};{nomProyecto};{responsable};{avance}\n"
            arch.write(registro)
        print(f"Entrega del Grupo {nroGrupo} guardada con éxito.")
    except OSError as error:
        print("Error al guardar la entrega en el archivo:", error)


def mostrar_entregas():
    """Muestra todas las entregas que tengan formato válido, omitiendo las corruptas."""
    print("\n=== LISTADO COMPLETO DE ENTREGAS VÁLIDAS ===")
    try:
        with open(RUTA_ENTREGAS, "rt", encoding="utf-8") as arch:
            esEncabezado = True
            hayRegistros = False
            for linea in arch:
                if esEncabezado:
                    esEncabezado = False
                else:
                    campos = linea.rstrip("\n").split(";")
                    if len(campos) == 4:
                        try:
                            nroGrupo = int(campos[0].strip())
                            proyecto = campos[1].strip()
                            responsable = campos[2].strip()
                            avance = float(campos[3].strip())

                            if nroGrupo > 0 and proyecto != "" and responsable != "" and 0.0 <= avance <= 100.0:
                                print(f"Grupo: {nroGrupo:2d} | Proyecto: {proyecto:15s} | Resp: {responsable:15s} | Avance: {avance:5.1f}%")
                                hayRegistros = True
                            else:
                                print(f"[Registro ignorado por inconsistencias: {linea.rstrip()}]")
                        except ValueError:
                            print(f"[Registro ignorado por valores inválidos: {linea.rstrip()}]")
                    else:
                        print(f"[Registro ignorado por cantidad errónea de campos: {linea.rstrip()}]")
            
            if not hayRegistros:
                print("No hay entregas válidas para mostrar.")
    except OSError as error:
        print("Error al leer el archivo de entregas:", error)


def buscar_entrega(nroGrupoBuscado):
    """Busca y muestra secuencialmente una entrega por su número de grupo."""
    print(f"\n--- BÚSQUEDA DEL GRUPO {nroGrupoBuscado} ---")
    encontrado = False
    try:
        with open(RUTA_ENTREGAS, "rt", encoding="utf-8") as arch:
            esEncabezado = True
            for linea in arch:
                if esEncabezado:
                    esEncabezado = False
                else:
                    campos = linea.rstrip("\n").split(";")
                    if len(campos) == 4:
                        try:
                            grupoActual = int(campos[0].strip())
                            if grupoActual == nroGrupoBuscado:
                                proyecto = campos[1].strip()
                                responsable = campos[2].strip()
                                avance = float(campos[3].strip())
                                print(f"¡Encontrado! Grupo {grupoActual} -> Proyecto: '{proyecto}' | Resp: {responsable} | Avance: {avance}%")
                                encontrado = True
                        except ValueError:
                            pass

        if not encontrado:
            print(f"No se encontró ninguna entrega registrada para el Grupo {nroGrupoBuscado}.")
    except OSError as error:
        print("Error al buscar en el archivo:", error)


def proyectos_avance_menor_50():
    """Informa los proyectos que tienen un porcentaje de avance inferior al 50%."""
    print("\n=== PROYECTOS CON AVANCE MENOR AL 50% ===")
    try:
        with open(RUTA_ENTREGAS, "rt", encoding="utf-8") as arch:
            esEncabezado = True
            hayMenores = False
            for linea in arch:
                if esEncabezado:
                    esEncabezado = False
                else:
                    campos = linea.rstrip("\n").split(";")
                    if len(campos) == 4:
                        try:
                            proyecto = campos[1].strip()
                            avance = float(campos[3].strip())
                            if avance < 50.0:
                                print(f"- Proyecto: {proyecto} (Avance: {avance:.1f}%)")
                                hayMenores = True
                        except ValueError:
                            pass
            
            if not hayMenores:
                print("Todos los proyectos registrados alcanzan o superan el 50% de avance.")
    except OSError as error:
        print("Error al filtrar proyectos:", error)


def calcular_promedio_avance():
    """Calcula el porcentaje promedio de avance protegiendo contra divisiones por cero."""
    sumaAvances = 0.0
    cantidadProyectos = 0

    try:
        with open(RUTA_ENTREGAS, "rt", encoding="utf-8") as arch:
            esEncabezado = True
            for linea in arch:
                if esEncabezado:
                    esEncabezado = False
                else:
                    campos = linea.rstrip("\n").split(";")
                    if len(campos) == 4:
                        try:
                            avance = float(campos[3].strip())
                            if 0.0 <= avance <= 100.0:
                                sumaAvances += avance
                                cantidadProyectos += 1
                        except ValueError:
                            pass

        print("\n=== ESTADÍSTICA GENERAL DE AVANCE ===")
        if cantidadProyectos > 0:
            promedio = sumaAvances / cantidadProyectos
            print(f"Promedio general de avance: {promedio:.2f}% (calculado sobre {cantidadProyectos} entregas válidas).")
        else:
            print("No se registraron entregas válidas para calcular el promedio (evitando división por cero).")
    except OSError as error:
        print("Error al calcular el promedio:", error)


def main():
    # 1. Creación del archivo de entregas
    crear_entregas_csv()

    # 2. Carga interactiva de entregas
    continuaPrograma = True
    while continuaPrograma:
        cargar_entrega()
        opcion = input("\n¿Desea ingresar otra entrega? (s/n): ").strip().lower()
        if opcion != "s":
            continuaPrograma = False

    # 3. Mostrar entregas
    mostrar_entregas()

    # 4. Buscar una entrega por número de grupo
    buscar_entrega(1)
    buscar_entrega(99)

    # 5. Proyectos con avance inferior a 50%
    proyectos_avance_menor_50()

    # 6. Promedio general
    calcular_promedio_avance()


main()
```
